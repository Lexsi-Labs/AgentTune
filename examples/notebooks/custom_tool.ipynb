# Custom BaseTool → GRPO

Train a LoRA policy that must call a **user** SQL tool. The warehouse lives in sqlite. The base model has never seen those rows.

Python can query sqlite. AgentTune is the other half: `BaseTool.to_schema()` is what the rollout engine advertises, `ToolRegistry.register_custom` is the name the spine can `get()`, and `create_agentic_trainer(..., tools=[tool])` builds AgentTune's `rollout_func` (not TRL native `tools=`). During generation it runs `execute`, inserts `role=tool` messages, then `reward_tool_used` / `reward_correct_answer` score the trace.

## 5  Custom tool

`QueryInventoryTool` is the AgentTune extension point: `name` / `description` / `_parameters()` / `execute() -> ToolResult`. Instantiation waits until sqlite exists (next cell). `tools=[tool]` later builds AgentTune's `rollout_func`, not TRL native `tools=`.


```python
class QueryInventoryTool(BaseTool):
    name = TOOL_NAME
    description = TOOL_DESCRIPTION

    def __init__(self, db_uri: str):
        self._sql = SQLDatabaseTool(db_uri)

    def _parameters(self):
        return {
            "type": "object",
            "properties": {
                "sql_command": {
                    "type": "string",
                    "description": "One read-only SELECT. Join customers/orders/order_lines/inventory as needed.",
                },
            },
            "required": ["sql_command"],
        }

    def execute(self, sql_command: str) -> ToolResult:
        out = self._sql.query(sql_command)
        if isinstance(out, dict) and out.get("error"):
            return ToolResult(success=False, output=None, error=str(out["error"]))
        return ToolResult(success=True, output=out)
```

## 6  Warehouse sqlite

Seed customers, orders, order_lines, inventory. None of these rows are in SmolLM2. The custom tool only reads this file.


```python
import sqlite3
from pathlib import Path

Path(DB_PATH).unlink(missing_ok=True)
conn = sqlite3.connect(DB_PATH)
conn.executescript("""
CREATE TABLE customers (
  id TEXT PRIMARY KEY, name TEXT NOT NULL, sla_hours INTEGER NOT NULL
);
CREATE TABLE orders (
  id TEXT PRIMARY KEY, customer_id TEXT NOT NULL, dest_dc TEXT NOT NULL, promised_date TEXT NOT NULL
);
CREATE TABLE order_lines (
  order_id TEXT NOT NULL, sku TEXT NOT NULL, qty_ordered INTEGER NOT NULL
);
CREATE TABLE inventory (
  sku TEXT PRIMARY KEY, name TEXT NOT NULL, qty INTEGER NOT NULL, price REAL NOT NULL, reorder_point INTEGER NOT NULL
);
""")
conn.executemany("INSERT INTO customers VALUES (?,?,?)", [
    ("CUST-ACME", "Acme Robotics", 24),
    ("CUST-SOL", "Solace AI", 48),
    ("CUST-NWD", "Northwind Labs", 12),
])
conn.executemany("INSERT INTO orders VALUES (?,?,?,?)", [
    ("ORD-1003", "CUST-ACME", "Dallas", "2026-09-03"),
    ("ORD-1008", "CUST-SOL", "Austin", "2026-09-04"),
    ("ORD-1011", "CUST-NWD", "Dallas", "2026-09-03"),
])
conn.executemany("INSERT INTO order_lines VALUES (?,?,?)", [
    ("ORD-1003", "SKU-C91", 10),
    ("ORD-1003", "SKU-A17", 4),
    ("ORD-1008", "SKU-H70", 6),
    ("ORD-1011", "SKU-F21", 25),
    ("ORD-1011", "SKU-G55", 40),
])
conn.executemany("INSERT INTO inventory VALUES (?,?,?,?,?)", [
    ("SKU-A17", "hex wrench 4mm", 42, 3.50, 10),
    ("SKU-B04", "M3 nylon washer", 180, 0.12, 50),
    ("SKU-C91", "24V DIN PSU", 7, 48.00, 8),
    ("SKU-D33", "Cat6 keystone", 64, 1.85, 20),
    ("SKU-E08", "ferrule 1.5mm", 500, 0.04, 100),
    ("SKU-F21", "DIN rail 35mm", 19, 6.40, 12),
    ("SKU-G55", "Wago 221-412", 96, 0.55, 30),
    ("SKU-H70", "PT100 probe", 11, 14.25, 6),
])
conn.commit()
conn.close()

sql_builtin = SQLDatabaseTool(DB_URI)
print("tables:", sql_builtin.list_tables())
print("schema:", sql_builtin.schema("inventory,order_lines,orders,customers"))
print("shortfall probe:", sql_builtin.query(
    "SELECT l.sku, l.qty_ordered, i.qty, (l.qty_ordered - i.qty) AS short "
    "FROM order_lines l JOIN inventory i ON i.sku=l.sku "
    "WHERE l.order_id='ORD-1003' AND l.qty_ordered > i.qty"
))
```

## 7  Register `QueryInventoryTool`

Subclass `BaseTool`, wrap `SQLDatabaseTool.query`, register under `TOOL_NAME`.


```python
tool = QueryInventoryTool(DB_URI)
ToolRegistry.register_custom(tool)
assert ToolRegistry.get(TOOL_NAME).name == TOOL_NAME

def query_inventory(sql_command: str):
    """Read-only SQL against Dallas DC: customers(id,name,sla_hours), orders(id,customer_id,dest_dc,promised_date), order_lines(order_id,sku,qty_ordered), inventory(sku,name,qty,price,reorder_point). One SELECT. No writes.

    Args:
        sql_command: A read-only SQL query to execute.

    Returns:
        Query results from QueryInventoryTool.execute.
    """
    result = tool.execute(sql_command=sql_command)
    if not result.success:
        return f"Error: {result.error}"
    return result.output

query_inventory.__name__ = TOOL_NAME
TOOLS = [query_inventory]
print("schema:", tool.to_schema())
print("execute:", tool.execute(sql_command="SELECT qty, price FROM inventory WHERE sku='SKU-A17'"))
print("trainer tool", TOOLS[0].__name__, "from BaseTool", tool.name)
```

## 8  Prompts

Ops tickets that need joins (order → lines → inventory / customer SLA). Gold is queried from the same sqlite the tool reads.


```python
conn = sqlite3.connect(DB_PATH)

def q(sql):
    return list(conn.execute(sql))

# Gold values come from the same sqlite the tool will query — not from a Python dict in execute().
short_1003 = q(
    "SELECT i.sku, (l.qty_ordered - i.qty) AS short, ROUND((l.qty_ordered - i.qty)*i.price, 2) AS gap "
    "FROM order_lines l JOIN inventory i ON i.sku=l.sku "
    "WHERE l.order_id='ORD-1003' AND l.qty_ordered > i.qty"
)
gap_1003 = short_1003[0][2] if short_1003 else 0
sku_short = short_1003[0][0] if short_1003 else ""
units_short = short_1003[0][1] if short_1003 else 0
below_rp = q("SELECT COUNT(*) FROM inventory WHERE qty < reorder_point")[0][0]
nwd_lines = q(
    "SELECT SUM(l.qty_ordered) FROM order_lines l "
    "JOIN orders o ON o.id=l.order_id WHERE o.customer_id='CUST-NWD'"
)[0][0]
sol_probe_onhand = q("SELECT qty FROM inventory WHERE sku='SKU-H70'")[0][0]
acme_sla = q("SELECT sla_hours FROM customers WHERE id='CUST-ACME'")[0][0]
psu_cover = q("SELECT qty FROM inventory WHERE sku='SKU-C91'")[0][0]
conn.close()

rows = [
    {
        "prompt": (
            "Ops ticket #4418 — Dallas is picking ORD-1003 for Acme Robotics this afternoon. "
            "Confirm whether every line can ship from on-hand stock. If any SKU is short, "
            "report the dollar gap at unit price (shortfall units × price). Number only."
        ),
        "answer": str(gap_1003),
    },
    {
        "prompt": (
            "Same ticket #4418 / ORD-1003. Which SKU is short against on-hand, and by how many units? "
            "Reply with the unit shortfall as a number (the SKU is for your SQL filter, not the reply)."
        ),
        "answer": str(units_short),
    },
    {
        "prompt": (
            "Replenishment board: how many SKUs are currently below reorder_point? "
            "This drives the buyer queue. Integer only."
        ),
        "answer": str(below_rp),
    },
    {
        "prompt": (
            "Northwind Labs (CUST-NWD) asked whether today's Dallas promise (ORD-1011) is oversized. "
            "Sum qty_ordered across their open lines. Integer only."
        ),
        "answer": str(nwd_lines),
    },
    {
        "prompt": (
            "Solace AI order ORD-1008 is six PT100 probes (SKU-H70). How many probes are on the shelf "
            "right now before we commit the pick? Integer only."
        ),
        "answer": str(sol_probe_onhand),
    },
    {
        "prompt": (
            "Acme's contract SLA is stored on customers.sla_hours. What is CUST-ACME's SLA in hours? "
            "Integer only."
        ),
        "answer": str(acme_sla),
    },
    {
        "prompt": (
            "Production wants another 24V DIN PSU (SKU-C91) for a line-down cell. "
            "On-hand qty after considering we already have a short on ORD-1003 is not required — "
            "report raw on-hand. Integer only."
        ),
        "answer": str(psu_cover),
    },
]
# Repeat the ticket set so MAX_SAMPLES still fills a GRPO epoch without inventing new gold.
while MAX_SAMPLES is not None and len(rows) < MAX_SAMPLES:
    rows.extend(rows[: max(1, MAX_SAMPLES - len(rows))])
if MAX_SAMPLES is not None:
    rows = rows[:MAX_SAMPLES]
print("train rows", len(rows))
print("sample prompt:\\n", rows[0]["prompt"])
print("sample gold", rows[0]["answer"])
```

## 9  Train

**Known caveats / gotchas:**

- `num_generations` must be ≥ 2.
- `per_device_train_batch_size` must be divisible by `num_generations` on current TRL.
- `tools=` builds agenttune's own `rollout_func` (`max_steps_per_turn`, `rollout_backend`, `engine_kwargs`). Native TRL `tools=` is not used.
- `reward_tool_used` scores a `role=tool` message in the rollout. `reward_correct_answer` needs an `answer` column.
- `peft_config` accepts a dict; the trainer builds `LoraConfig` from it.


```python
peft_config = None
if USE_PEFT:
    peft_config = {
        "r": LORA_R,
        "lora_alpha": LORA_ALPHA,
        "lora_dropout": LORA_DROPOUT,
        "bias": LORA_BIAS,
        "task_type": TASK_TYPE,
        **({"target_modules": LORA_TARGETS} if LORA_TARGETS else {}),
    }

common = dict(
    algorithm                    = "grpo",
    backend                      = BACKEND,
    model                        = MODEL,
    dataset_name                 = DATASET,
    dataset_config               = DATASET_CONFIG,
    split                        = SPLIT,
    max_samples                  = MAX_SAMPLES,
    column_mapping               = COLUMN_MAPPING,
    system_prompt                = SYSTEM_PROMPT,
    enable_thinking              = ENABLE_THINKING,
    reward_funcs                 = REWARD_FUNCS,
    reward_weights               = REWARD_WEIGHTS,
    tools                        = TOOLS,
    rollout_backend              = ROLLOUT_BACKEND,
    max_steps_per_turn           = MAX_STEPS_PER_TURN,
    engine_kwargs                = ENGINE_KWARGS,
    peft_config                  = peft_config,
    output_dir                   = OUT_DIR,
    num_train_epochs             = NUM_EPOCHS,
    # max_steps                  = MAX_STEPS,  # commented: 1 epoch over MAX_SAMPLES (20–100)
    per_device_train_batch_size  = BATCH_SIZE,
    gradient_accumulation_steps  = GRAD_ACCUM,
    learning_rate                = LR,
    warmup_ratio                 = WARMUP_RATIO,
    warmup_steps                 = WARMUP_STEPS,
    weight_decay                 = WEIGHT_DECAY,
    lr_scheduler_type            = LR_SCHEDULER,
    max_grad_norm                = MAX_GRAD_NORM,
    seed                         = SEED,
    data_seed                    = DATA_SEED,
    optim                        = OPTIMIZER,
    num_generations              = NUM_GENERATIONS,
    max_prompt_length            = MAX_PROMPT_LENGTH,
    max_completion_length        = MAX_COMPLETION_LENGTH,
    temperature                  = TEMPERATURE,
    top_p                        = TOP_P,
    beta                         = BETA,
    use_vllm                     = USE_VLLM,
    vllm_mode                    = VLLM_MODE,
    log_completions              = LOG_COMPLETIONS,
    chat_template_kwargs         = CHAT_TEMPLATE_KWARGS,
    logging_steps                = LOGGING_STEPS,
    save_steps                   = SAVE_STEPS,
    eval_strategy                = EVAL_STRATEGY,
    report_to                    = REPORT_TO,
    push_to_hub                  = PUSH_TO_HUB,
    bf16                         = BF16,
    fp16                         = FP16,
    gradient_checkpointing       = GRAD_CKPT,
    dataloader_num_workers       = DATALOADER_WORKERS,
    remove_unused_columns        = REMOVE_UNUSED_COLUMNS,
)
common["train_dataset"] = Dataset.from_list(rows)
common.pop("dataset_name", None)

trainer = create_agentic_trainer(**common)
```
